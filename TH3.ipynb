# Техническое задание: TenderHack3

## Общее описание
Разработка платформы для управления товарами, закупками и группами товаров с интеллектуальной системой поиска и рекомендаций.

## Функциональные требования

### 1. Система авторизации и регистрации
- Регистрация новых пользователей
- Привязка пользователей к CNN
- Авторизация с использованием существующих данных из таблицы закупок

### 2. Структура фронтенда
**Основные страницы:**
- Товары
- Закупки 
- Группы товаров

### 3. Поисковая система
- Универсальная поисковая строка на всех страницах
- Маршрутизация запросов к соответствующим endpoint'ам в зависимости от текущей страницы:
  - Товары → endpoint товаров
  - Закупки → endpoint закупок  
  - Группы → endpoint групп
- Интеграция с ML-системой для интеллектуального поиска

### 4. Страница товаров
- Поисковая строка
- Базовый список рекомендуемых товаров (2-3 позиции)
- Боковая панель с категориями:
  - 10 самых популярных категорий
  - Автоматическая генерация категорий (аналогично TH2)
- Карточки товаров с кнопками:
  - "Добавить в избранное"
  - "Добавить в корзину" 
  - "Добавить в группу" (открывает чекбокс-окно со списком групп)

### 5. Страница закупок
- Поисковая строка для закупок
- Список закупок из БД
- Кнопка "Создать на основании" / "Повторить закупку"
- При нажатии "Создать на основании" - товары из закупки отправляются в корзину закупок
- Отсутствующие товары подсвечиваются серым и блокируют оформление

### 6. Система корзин
**Корзина закупок:**
- Редактирование каждой закупки
- Отмена запросов из корзины

### 7. Система групп товаров
- Создание пользовательских групп
- ML-генерация групп на основе поисковых запросов
- Пример: запрос "Оборудование СТО" → автоматическое создание группы с соответствующими товарами

### 8. Типовые закупки
- Раздел "Категории" с популярными категориями
- Основано на типовых закупках от жюри
- Создание профиля типовой закупки
- Название формируется на основе типа организации или группы товаров
- Возможность создания пользовательских типовых закупок

## Технические требования

### Backend
- Раздельные endpoint'ы для товаров, закупок, групп
- Интеграция с ML-системой
- Экспертная модель для рекомендаций

### База данных
- Адаптация под предоставленную структуру данных
- Оптимизация для поисковых запросов

### Интерфейс
- Единая стилистика блоков
- Адаптивная верстка
- Интуитивная навигация

## Бизнес-логика
- Автоматическая генерация категорий товаров
- Интеллектуальный подбор товаров в группы
- Валидация доступности товаров при создании закупок
- Система рекомендаций на основе пользовательской активности
